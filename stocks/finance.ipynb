```python
import yfinance as yf
import pandas as pd
from pandas_datareader import data as pdr
import datetime as dt
import numpy as np
```

```python
end_date = dt.datetime.now()

start_date = end_date - dt.timedelta(days = 365*5)
end_date, start_date
```

```python
stocks = ['MSFT','SPY','QQQ']
```

```python
df = yf.download(stocks, start = start_date, end = end_date, auto_adjust = False)
```

```python
df.head()
```

```python
adj_close_price = df['Adj Close']
adj_close_price.head()
```

```python
log_returns = np.log(adj_close_price/adj_close_price.shift(1))
log_returns.head()
```

```python
cumulative_log_returns = log_returns.cumsum()
cumulative_log_returns.head()
```

```python
cumulative_log_returns.plot(title = "Cumulative Returns", figsize = (10,6))
```
