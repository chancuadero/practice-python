```python
transactions = [
    {
        "symbol": "BTC",
        "type": "buy",
        "quantity": 0.7,
        "price": 27090.4,
        "date": "2022-03-07",
    },
    {
        "symbol": "BTC",
        "type": "sell",
        "quantity": 0.1,
        "price": 26561.4,
        "date": "2022-04-12",
    },
    {
        "symbol": "BTC",
        "type": "buy",
        "quantity": 0.4,
        "price": 28090.4,
        "date": "2022-05-21",
    },
    {
        "symbol": "BTC",
        "type": "buy",
        "quantity": 0.3,
        "price": 25090.4,
        "date": "2022-06-02",
    },
    {
        "symbol": "BTC",
        "type": "sell",
        "quantity": 0.5,
        "price": 32034.67,
        "date": "2022-06-29",
    },
    {
        "symbol": "ETH",
        "type": "buy",
        "quantity": 6,
        "price": 1754,
        "date": "2022-05-21",
    },
    {
        "symbol": "ETH",
        "type": "buy",
        "quantity": 2,
        "price": 1950,
        "date": "2022-06-02",
    },
    {
        "symbol": "ETH",
        "type": "sell",
        "quantity": 3,
        "price": 2100.67,
        "date": "2022-06-29",
    },
]
```

From this list, using a loop, create a dictionary with keys for each of the cryptocurrencies present in the transactions and an empty dictionary as its value. 

```python
portfolio = {}
for transaction in transactions:
    symbol = transaction['symbol']
    if symbol not in portfolio:
        portfolio[symbol] = {}

print(portfolio)
```

1. From this list, using a loop, add to the previous dictionary a key "quantity" which has as its value the quantity still present in the portfolio. (You will need to use conditions to check if these are purchases or sales).

```python
for transaction in transactions:
    symbol = transaction['symbol']
    if 'quantity' not in portfolio[symbol]:
        portfolio[symbol]['quantity'] = 0
    if transaction['type'] == 'buy':
        portfolio[symbol]['quantity'] += transaction['quantity']
    elif transaction['type'] == 'sell':
        portfolio[symbol]['quantity'] -= transaction['quantity']

print(portfolio)
```

2. Considering today's BTC price of 35000 and that of ETH of 2400, add to our previous dictionary our PNL (i.e., our profit or our loss) for each of the cryptos, with the key pnl.

```python
current_prices = {'BTC': 35000, 'ETH': 2400}

for symbol in portfolio:
    total_cost = sum(
        t['quantity'] * t['price']
        for t in transactions
        if t['symbol'] == symbol and t['type'] == 'buy'
    )

    current_value = (
        portfolio[symbol]['quantity'] * current_prices[symbol]
    )

    portfolio[symbol]['pnl'] = round(current_value - total_cost,2)

print(portfolio)
```

3. Add to the dictionary the average purchase price for each of the cryptos.

```python
for symbol in portfolio:
    total_purchase = sum(
        t['quantity']
        for t in transactions
        if t['symbol'] == symbol and t['type'] == 'buy'
    )

    total_purchase_cost = sum(
        t['quantity'] * t['price']
        for t in transactions
        if t['symbol'] == symbol and t['type'] == 'buy'
    )

    portfolio[symbol]['avg_purchase_price'] = round(total_purchase_cost / total_purchase,2)

print(portfolio)
```

4. (Bonus) For each of the cryptos: display the quantity purchased after the date of 2022-03-02. For this, you will need to find out how to interact with dates; you need to use the datetime library.

```python
from datetime import datetime

for symbol in portfolio:
    total_purchase = sum(
        t['quantity']
        for t in transactions
        if t['symbol'] == symbol and t['type'] == 'buy' and datetime.strptime(t['date'],"%Y-%m-%d") > datetime.strptime("2022-03-02","%Y-%m-%d")
    )
    portfolio[symbol]['total_purchase_after_june_2022'] = total_purchase


print(portfolio)
```
