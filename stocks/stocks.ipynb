John is an investor with Apple stocks. Each Apple stock is currently valued at $135. He bought 20 stocks at $60 each, then acquired an additional 10 at $145 each. He sold 8 stocks at $120 each. Now, John is evaluating the financial merits of selling his remaining stocks and is curious about the potential tax implications.


Tasks:
1. Calculate and display the total value of the current Apple stocks in John's portfolio.
2. Calculate and display the average purchase price of John's Apple stocks.
3. Calculate and display the potential total capital gain if John sells all his Apple stocks now.
4. John holds dual citizenship in France and Belguim. In France, the tax rate on capital gains is 30%. In Belgium, capital gains are tax-exempt up to $500, and a 40% tax rate applies beyond this threshold. Calculate and display the taxes John would incur on his capital gains in France and Belgium.

```python
#Task 1
current_market_price = 135
apple_shares1 = 20
apple_shares2 = 10

apple_purchase_price1 = 60
apple_purchase_price2 = 145

stocks_sold = 8
stocks_sold_price = 120

total_owned_shares = (apple_shares1 + apple_shares2) - stocks_sold

total_value =  (total_owned_shares) * current_market_price
print(f"The total market value of the Apple stock in John's portfolio is ${total_value}")
```

```python
#Task 2
transaction_1 = apple_shares1 * apple_purchase_price1
transaction_2 = apple_shares2 * apple_purchase_price2
total_investment = transaction_1 + transaction_2
total_shares = apple_shares1 + apple_shares2
avg_purchase_price = total_investment / total_shares
print(f"The average purchase price of John's Apple stock is ${avg_purchase_price:.2f} per share")
```

```python
#Task 3
total_first_capital_gain = (apple_shares1 - stocks_sold) * (current_market_price - apple_purchase_price1)
total_second_capital_gain = apple_shares2 * (current_market_price-apple_purchase_price2)
sold_stocks_capital_gain = (stocks_sold_price - apple_purchase_price1) * stocks_sold
capital_gain = total_first_capital_gain + total_second_capital_gain + sold_stocks_capital_gain
print(f"Potential capital gain for John's Apple shares: ${capital_gain}")
```

```python
#Task 4
tax_owed_france = capital_gain * 0.3
tax_owed_belgium = (capital_gain - 500) * .40
print(f"John owes a tax of ${tax_owed_france} in France, while in Belgium the tax he owes is ${tax_owed_belgium}")
```
